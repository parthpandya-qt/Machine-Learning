```python
companyName=[]
rating=[]
review=[]
for i in range(1,11):
    headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    data = requests.get(f'https://www.ambitionbox.com/list-of-companies?page={i}',headers = headers,timeout=10)
    soup = BeautifulSoup(data.text, 'lxml')
    
    for i in soup.find_all("div",class_='companyCardWrapper__primaryInformation'):
        companyName.append(i.find("h2",class_='companyCardWrapper__companyName').text.strip())
        rating.append(i.find("div", class_="rating_text rating_text--md").text.strip())
        review.append(i.find("span",class_="companyCardWrapper__companyRatingCount").text.strip())
    d={'name':companyName, 'rating':rating, 'review':review}
    df = pd.DataFrame(d)
    
        
    


    
```
