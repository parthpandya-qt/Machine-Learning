```python
# encoding numerical features to categorical data 
#type discretization and binarization
#Disretization(binning) :  process to transform continuous variable into discrete variable by creating a set of contiguous intervals that span the range of the variables values.
# why use it: 1. to reduce the effect of outliers by grouping them into bins
            # 2. to capture non-linear relationships between the feature and the target variable by creating bins
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived']  )
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
x = df.iloc[:,1:]
y = df.iloc[:,0]
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.2,random_state=42)
```

```python
x_train.head()
```

```python
kbinAge = KBinsDiscretizer(n_bins=5,encode='ordinal',strategy='uniform')
kbinFare = KBinsDiscretizer(n_bins=5,encode='ordinal',strategy='quantile')
```

```python
trf = ColumnTransformer([
    ('binAge',kbinAge,[0]),
    ('binFare',kbinFare,[1])
])
```

```python
x_train_trf = trf.fit_transform(x_train)
x_test_trf = trf.transform(x_test)
```
