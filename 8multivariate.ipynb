```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
titanic = pd.read_csv('train.csv')
iris = sns.load_dataset('iris')
flights = sns.load_dataset('flights')
```

```python
#scatterPlot (Numerical-Numerical)
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])

```

```python
#multivariate using scatter plot
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'])
```

```python
# barplot (Numerical - categorical)
sns.barplot(x=titanic['Pclass'],y=titanic['Fare'],hue=titanic['Sex'])
```

```python
#box plot  (Numerical - categorical)
sns.boxplot(x=titanic['Sex'],y=titanic['Age'] , hue = titanic['Survived'])
```

```python
sns.histplot(x=titanic['Age'], kde=True)
```

```python
sns.histplot(titanic[titanic['Survived']==0]['Age'], kde=True)
sns.histplot(titanic[titanic['Survived']==1]['Age'], kde=True)
```

```python
sns.kdeplot(titanic[titanic['Survived']==0]['Age'], label='Did Not Survive')
sns.kdeplot(titanic[titanic['Survived']==1]['Age'], label='Survived')

plt.legend()
plt.show() 
```

```python
#heataMap    categorical - categorical
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Sex').mean(numeric_only=True)['Survived'] * 100).plot(kind='bar') 
```

```python
#clusterMap (categorical-categorical)
sns.clustermap(
    pd.crosstab(titanic['Parch'], titanic['Survived'])
)
```

```python
sns.pairplot(iris)
```
