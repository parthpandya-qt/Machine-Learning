```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('train.csv',usecols = ['Age','Fare','Survived'])
```

```python
df.head()
```

```python
df.isnull().mean() * 100
```

```python
x = df.drop('Survived', axis=1) 
y = df['Survived']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
x_train.head()
```

```python
x_train['Age_imputed'] = x_train['Age']
x_test['Age_imputed'] = x_test['Age'] 
```

```python
# Preserves the variance of the original data by imputing missing values with the mean of the training set.
# Memory heavy for the deployment.
# well suited for linear models and not for tree based models.
# covariance get distorted when we impute the missing values with the random imputer.
```

```python
x_train['Age_imputed'][x_train['Age_imputed'].isnull()] = \
    x_train['Age'].dropna().sample(x_train['Age'].isnull().sum()).values

x_test['Age_imputed'][x_test['Age_imputed'].isnull()] = \
    x_train['Age'].dropna().sample(x_test['Age'].isnull().sum()).values
```
