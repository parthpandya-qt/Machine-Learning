```python
cols = [var for var in df.columns if df[var].isnull().mean()<.05 and df[var].isnull().mean()>0]
```

```python
cols
```

```python
df[cols].head()
```

```python
len(df[cols].dropna())/len(df)   #percentage of data that will be retained after dropping the missing values in the selected columns
```

```python
new_df = df[cols].dropna()
```

```python
df.shape
```

```python
new_df.shape
```

```python
temp = pd.concat([
    df['enrolled_university'].value_counts()/len(df),
    new_df['enrolled_university'].value_counts()/len(new_df)
],axis = 1)
temp.columns = ['original', 'new']
```

```python
temp = pd.concat([
    df['education_level'].value_counts()/len(df),
    new_df['education_level'].value_counts()/len(new_df)
],axis = 1)
temp.columns = ['original', 'new']
```
