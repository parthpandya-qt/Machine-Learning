```python
# Steps of MICE
# Initialize missing values
# Fill all missing values with simple estimates:
# Mean (numerical)
# Mode (categorical)
# Median, etc.
# Select one variable with missing values
# Suppose column A has missing values.
# Treat A as the target variable.
# Use all other columns (B, C, D, ...) as predictors.
# Train a model
# Use rows where A is not missing.
# Fit a regression/classification model:
# Linear Regression for continuous data
# Logistic Regression for binary data
# Other suitable models
# Predict missing values
# Predict the missing values of A using the trained model.
# Replace the initial guesses with these predictions.
# Move to the next variable
# Repeat Steps 2–4 for column B, then C, etc.
# Each variable is imputed using the most recently updated values of the other variables.
# Complete one iteration
# After all variables have been imputed once, one cycle (iteration) is completed.
# Repeat multiple iterations
# Continue cycling through all variables.
# The imputations gradually stabilize.
# Generate multiple datasets (optional)
# Classical MICE creates several imputed datasets (e.g., 5–10).
# Analysis is performed on each dataset.
# Results are combined to account for uncertainty.
```

```python
import pandas as pd
import numpy as np

# Sample dataset
df = pd.read_csv('train.csv')[['Age','Pclass','Fare','Survived']]

print("Original Data:")
print(df)

# Enable IterativeImputer (MICE)
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Create imputer
mice_imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

# Fit and transform
df_imputed = pd.DataFrame(
    mice_imputer.fit_transform(df),
    columns=df.columns
)

print("\nImputed Data:")
print(df_imputed)
```
